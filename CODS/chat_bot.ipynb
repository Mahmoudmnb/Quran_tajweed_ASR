```python
questions = [
    "كيف أسجل في الجامعة؟",
    "ما هي الكليات الموجودة؟"
]

answers = [
    "يمكنك التسجيل من خلال...",
    "توجد كليات مثل الطب والهندسة..."
]
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

q_embeddings = model.encode(questions)
```

```python
user_q = "شلون بدي قدم؟"
user_emb = model.encode([user_q])
```

```python
import numpy as np

scores = np.dot(q_embeddings, user_emb.T).squeeze()
best_idx = scores.argmax()
```

```python
print(answers[best_idx])
```
