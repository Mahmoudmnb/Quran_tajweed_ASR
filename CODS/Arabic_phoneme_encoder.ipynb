```python
def remove_quranic_marks(parsed_text):
    """
    Remove all non-phonetic Quranic symbols.
    """

    return [(l, d) for l, d in parsed_text if l not in QURANIC_MARKS]
```

```python
def parse_arabic(text: str) -> List[Tuple[str, List[str]]]:
    """
    Parse Arabic text into (letter, diacritics) pairs.

    Example:
        "مِنْ" → [("م", ["ِ"]), ("ن", ["ْ"])]

    Rules:
    - Each base letter is grouped with its following diacritics
    - Diacritics are stored as a list
    - Spaces are preserved as (" ", [])
    - Quranic marks are ignored
    """

    result: List[Tuple[str, List[str]]] = []

    current_letter = None
    current_diacritics: List[str] = []

    for i, char in enumerate(text):

        # -------------------------------
        # 1. Skip Quranic marks
        # -------------------------------
        if char in QURANIC_MARKS:
            continue

        # ا۟  skip this letter like كَانُوا۟

        if (
            i + 1 < len(text)
            and char == "ا"
            and (text[i + 1] == "۟" or text[i + 1] == "۠")
        ):
            continue

        # -------------------------------
        # 2. Handle space
        # -------------------------------
        if char.isspace():
            # flush previous letter
            if current_letter is not None:
                result.append((current_letter, current_diacritics))
                current_letter = None
                current_diacritics = []

            result.append((" ", []))
            continue

        # -------------------------------
        # 3. If diacritic → attach it
        # -------------------------------
        if char in DIACRITICS:
            if current_letter is None:
                # edge case: diacritic without base letter
                continue

            current_diacritics.append(char)
            continue

        # -------------------------------
        # 4. New letter → flush previous
        # -------------------------------
        if current_letter is not None:
            result.append((current_letter, current_diacritics))

        current_letter = char
        current_diacritics = []

    # -------------------------------
    # 5. Flush last letter
    # -------------------------------
    if current_letter is not None:
        result.append((current_letter, current_diacritics))

    return result
```

```python
def is_madd_letter(letter, diacritics):
    """
    Check whether a letter can function as a madd letter.

    Madd letters:
    - ا
    - وْ
    - يْ

    They must not carry real vowels.
    """

    # Alif normally has no diacritics
    if letter == LONG_ALIF:
        return len(diacritics) == 0

    # Waw and Ya may carry sukoon
    if letter in [LONG_WAW, LONG_YA]:

        # Allowed:
        # []
        # [ْ]
        return len(diacritics) == 0 or diacritics == [SUKOON]

    return False


def is_word_final(parsed_text, i):
    j = i + 1
    while j < len(parsed_text):
        if parsed_text[j][0] == SPACE_TOKEN:
            return True

        if parsed_text[j][0].isalpha():
            return False
        j += 1
    return True


def parsed_to_string(parsed):

    text = ""

    for letter, diacritics in parsed:
        text += letter
        text += "".join(diacritics)

    return text


def has_vowel(diacritics):
    """
    Check whether a letter carries a real vowel.
    """

    return any(
        d
        in [
            FATHA,
            DAMMA,
            KASRA,
            FATHATAN,
            DAMMATAN,
            KASRATAN,
        ]
        for d in diacritics
    )
```

```python
def normalize_dagger_alif(parsed_text):
    """
    Convert dagger alif into LONG_ALIF marker.

    Example:
        مَـٰ → م + LONG_ALIF
    """

    result = []

    for letter, diacritics in parsed_text:

        if DAGGER_ALIF in diacritics or TATWEEL in diacritics:

            # remove dagger alif
            diacritics = [d for d in diacritics if d != DAGGER_ALIF]

            # add normalized long vowel marker
            diacritics.append("LONG_ALIF")

        result.append((letter, diacritics))

    return result
```

```python
def normalize_alif_maqsura(parsed_text):
    """
    Resolve final alif maqsurah (ى).

    Rules:

    1. Previous fatha + no vowel on ى
       -> LONG_ALIF

    2. Previous kasra + no vowel on ى
       -> LONG_YA

    3. ىً
       -> tanween fatḥ ("an")

    4. Otherwise
       -> consonant yaa
    """

    result = []

    for i, (letter, diacritics) in enumerate(parsed_text):

        # -----------------------------------
        # Normal letters
        # -----------------------------------
        if letter != "ى":
            result.append((letter, diacritics))
            continue

        # -----------------------------------
        # Need previous letter
        # -----------------------------------
        if len(result) == 0:
            result.append(("ي", diacritics))
            continue

        prev_letter, prev_diacritics = result[-1]

        # -----------------------------------
        # Case 1:
        # ىً -> an
        # -----------------------------------
        if FATHATAN in diacritics:

            prev_diacritics.append(FATHATAN)
            result[-1] = (prev_letter, prev_diacritics)

            continue

        # -----------------------------------
        # Detect whether current ى
        # has real vowel
        # -----------------------------------
        has_real_vowel = any(
            d in diacritics
            for d in [
                FATHA,
                DAMMA,
                KASRA,
                DAMMATAN,
                KASRATAN,
            ]
        )

        # -----------------------------------
        # Case 2:
        # previous fatha -> aa
        # -----------------------------------
        if LONG_ALIF_MARK in diacritics or (
            FATHA in prev_diacritics
            and not has_real_vowel
            and is_word_final(parsed_text, i)
            and SUKOON not in diacritics
        ):

            prev_diacritics.append(LONG_ALIF_MARK)
            result[-1] = (prev_letter, prev_diacritics)

            continue

        # -----------------------------------
        # Case 3:
        # previous kasra -> ii
        # -----------------------------------
        if (
            KASRA in prev_diacritics
            and not has_real_vowel
            and is_word_final(parsed_text, i)
        ):

            prev_diacritics.append(LONG_YA_MARK)
            result[-1] = (prev_letter, prev_diacritics)

            continue

        # -----------------------------------
        # Otherwise:
        # treat as consonant yaa
        # -----------------------------------
        result.append(("ي", diacritics))

    return result
```

```python
def apply_long_vowels(parsed_text):
    """
    Modify parsed text to merge long vowel patterns.

    Input:
        [("ق", ["َ"]), ("ا", []), ("ل", ["َ"])]

    Output:
        [("ق", ["َ", "LONG_ALIF"]), ("ل", ["َ"])]

    Then Step 2 will convert it to:
        ["q", "aa", "l", "a"]
    """

    result = []
    i = 0

    while i < len(parsed_text):
        letter, diacritics = parsed_text[i]

        # ----------------------------------
        # Look ahead for long vowel patterns
        # ----------------------------------
        if i + 1 < len(parsed_text):
            next_letter, next_diacritics = parsed_text[i + 1]

            # ------------------------------
            # 1. Fatha + Alif → aa
            # ------------------------------
            if (
                FATHA in diacritics
                and next_letter == LONG_ALIF
                and is_madd_letter(next_letter, next_diacritics)
            ):
                result.append((letter, diacritics + ["LONG_ALIF"]))
                i += 2
                continue

            # ------------------------------
            # 2. Kasra + Ya → ii
            # ------------------------------
            if (
                KASRA in diacritics
                and next_letter == LONG_YA
                and is_madd_letter(next_letter, next_diacritics)
            ):
                result.append((letter, diacritics + ["LONG_YA"]))
                i += 2
                continue

            # ------------------------------
            # 3. Damma + Waw → uu
            # ------------------------------
            if (
                DAMMA in diacritics
                and next_letter == LONG_WAW
                and is_madd_letter(next_letter, next_diacritics)
            ):
                result.append((letter, diacritics + ["LONG_WAW"]))
                i += 2
                continue

        # ----------------------------------
        # No long vowel → keep as is
        # ----------------------------------
        result.append((letter, diacritics))
        i += 1

    return result
```

```python
def apply_al_rule(parsed_text):
    """
    Apply sun/moon lam rules.

    Sun letters:
        lam disappears
        next letter gets shadda

    Moon letters:
        no change
    """

    result = []
    i = 0

    while i < len(parsed_text):

        # Need at least 3 chars:
        # ٱ + ل + next letter
        if i + 2 < len(parsed_text):

            first_letter, first_diacritics = parsed_text[i]
            second_letter, second_diacritics = parsed_text[i + 1]
            third_letter, third_diacritics = parsed_text[i + 2]

            # -----------------------------------
            # Detect الـ
            # -----------------------------------
            if (
                first_letter == HAMZAT_WASL
                and second_letter == "ل"
                and SUKOON not in second_diacritics
            ):

                # -----------------------------------
                # Sun letter case
                # -----------------------------------
                if third_letter in SUN_LETTERS:

                    # add shadda if missing
                    if SHADDA not in third_diacritics:
                        third_diacritics = [SHADDA] + third_diacritics

                    result.append((first_letter, first_diacritics))
                    result.append((third_letter, third_diacritics))

                    i += 3
                    continue

        # -----------------------------------
        # default
        # -----------------------------------
        result.append(parsed_text[i])
        i += 1

    return result
```

```python
def apply_madd_silah(parsed_text, next_word):
    """
    Apply Madd Silah Sughra.

    هُ → huu
    هِ → hii

    when followed by a vowelized letter.
    """

    result = []
    prev_diac = None

    str_text = parsed_to_string(parsed_text)

    #     Processing word: فِيهِ
    #     Processing word: مُهَانًا

    for i, (letter, diacritics) in enumerate(parsed_text):

        # -----------------------------------
        # Must have next letter
        # -----------------------------------

        if len(next_word) > 0:

            _, next_diacritics = next_word[0]

            next_text = parsed_to_string(next_word)

            if (
                has_vowel(next_diacritics)
                and is_word_final(parsed_text, i)
                and prev_diac is not None
                and len(prev_diac) > 0
                and SUKOON not in prev_diac
            ):

                # -----------------------------------
                # هُ → huu
                # -----------------------------------
                if (
                    letter == "ه"
                    and DAMMA in diacritics
                    and (str_text != "يَرْضَهُ" and next_text != "لَكُمْ")
                ):

                    diacritics = diacritics + ["LONG_WAW"]

                # -----------------------------------
                # هِ → hii
                # -----------------------------------
                elif letter == "ه" and KASRA in diacritics:

                    diacritics = diacritics + ["LONG_YA"]

            elif letter == "ه" and str_text == "فِيهِ" and next_text == "مُهَانًا":
                diacritics = diacritics + ["LONG_YA"]

        result.append((letter, diacritics))

        prev_diac = diacritics

    return result
```

```python
def arabic_to_phonemes(parsed_text):
    """
    Convert parsed Arabic [(letter, diacritics)] → list of phonemes.

    Output:
        ["m", "i", "n", "b", "a", "ʕ", "d", "i"]
    """

    phonemes = []
    word_closed = False

    for i, (letter, diacritics) in enumerate(parsed_text):

        if word_closed:
            continue

        # -------------------------------
        # 1. Handle space
        # -------------------------------
        if letter == " ":

            phonemes.append(SPACE_TOKEN)
            word_closed = False
            continue

        # -------------------------------
        # 2. Get base consonant
        # -------------------------------
        base = ARABIC_TO_PHONEME.get(letter)

        if base is None:
            # unknown character → skip
            continue

        # -------------------------------
        # 3. Check shadda
        # -------------------------------
        has_shadda = SHADDA in diacritics

        # double consonant
        if has_shadda:

            # this is for words like وَٱلنَّبِيِّـۧنَ so we have madd after shadda

            if (
                letter == LONG_YA
                and i + 1 < len(parsed_text)
                and KASRA in parsed_text[i - 1][1]
                and KASRA in diacritics
            ):
                diacritics.append(LONG_YA_MARK)

            if base == "n" or base == "m":
                phonemes.append(base + base)
            else:
                phonemes.append(base)

        # always append base once
        phonemes.append(base)

        # -------------------------------
        # Detect long vowel first
        # -------------------------------
        has_long_alif = LONG_ALIF_MARK in diacritics
        has_long_ya = LONG_YA_MARK in diacritics
        has_long_waw = LONG_WAW_MARK in diacritics

        # -------------------------------
        # Emit vowel
        # -------------------------------
        for d in diacritics:

            # 🔴 Skip short vowel if long exists
            if has_long_alif and d == FATHA:
                continue

            if has_long_ya and d == KASRA:
                continue

            if has_long_waw and d == DAMMA:
                continue

            # -------------------------------
            # Normal handling
            # -------------------------------
            if d == FATHA:
                phonemes.append("a")

            elif d == DAMMA:
                phonemes.append("u")

            elif d == KASRA:
                phonemes.append("i")

            elif d == FATHATAN:
                phonemes.append("an")
                word_closed = True

            elif d == DAMMATAN:
                phonemes.append("un")
                word_closed = True

            elif d == KASRATAN:
                phonemes.append("in")
                word_closed = True

            elif d == LONG_ALIF_MARK:
                phonemes.append("aa")

            elif d == LONG_YA_MARK:
                phonemes.append("ii")

            elif d == LONG_WAW_MARK:
                phonemes.append("uu")

            elif d == SUKOON:
                pass

            elif d == SHADDA:
                pass

    return phonemes
```

```python
def apply_hamzaat_wasl(parsed: list) -> list:

    str_text = parsed_to_string(parsed)

    if str_text == "ٱمْرُؤٌ۟":
        parsed[0] = (HAMZAT_WASL, [DAMMA])
        return parsed
    elif str_text in ["ٱبْنُ", "وَٱبْنَ", "ٱبْنَ", "ٱبْنَتَىَّ", "ٱبْنَتَىَّ"]:
        parsed[0] = (HAMZAT_WASL, [KASRA])
        return parsed

    if str_text.__contains__(HAMZAT_WASL):

        i = 0

        while i < len(parsed):
            if parsed[i][0] == HAMZAT_WASL:
                if i + 1 < len(parsed) and parsed[i + 1][0] == "ل":
                    parsed[i] = (HAMZAT_WASL, [FATHA])
                    break
                else:
                    if i + 2 < len(parsed) and DAMMA in parsed[i + 2][1]:
                        parsed[i] = (HAMZAT_WASL, [DAMMA])
                        break
                    else:
                        parsed[i] = (HAMZAT_WASL, [KASRA])
                        break
            i += 1
    return parsed
```

```python
def apply_sukoon_noon_and_tanween_rules(phonemes: list) -> list:

    final_phonemes = []

    for i, phoneme in enumerate(phonemes):

        # handle idgham rules for noon and tanween .
        if (
            phoneme in ["n", "an", "un", "in"]
            and i + 3 < len(phonemes)
            and phonemes[i + 1] == SPACE_TOKEN
            and phonemes[i + 2] in IDGHAM_LETTERS
            and (
                phonemes[i + 3] in ["a", "u", "i", "aa", "uu", "ii"]
                or i + 4 < len(phonemes)
                and phonemes[i + 2][0] == phonemes[i + 3][0]
                and phonemes[i + 4] in ["a", "u", "i", "aa", "uu", "ii"]
            )
        ):

            if phoneme == "an":
                final_phonemes.append("a")
            elif phoneme == "un":
                final_phonemes.append("u")
            elif phoneme == "in":
                final_phonemes.append("i")

            if (
                i + 3 < len(phonemes)
                and phonemes[i + 2] == phonemes[i + 3]
                or phonemes[i + 2] in ["nn", "mm"]
            ):
                continue
            else:
                if phonemes[i + 2] in ["l", "r"]:
                    final_phonemes.append(phonemes[i + 2])
                else:
                    final_phonemes.append(phonemes[i + 2] + phonemes[i + 2])

        # apply iqlab rules for noon and tanween before .
        elif phoneme in ["n", "an", "un", "in"] and (
            (
                i + 3 < len(phonemes)
                and phonemes[i + 1] == SPACE_TOKEN
                and phonemes[i + 2] in IKLAAB_LETTERS
                and phonemes[i + 3] in ["a", "u", "i", "aa", "uu", "ii"]
            )
            or (
                i + 2 < len(phonemes)
                and phonemes[i + 1] in IKLAAB_LETTERS
                and phonemes[i + 2] in ["a", "u", "i", "aa", "uu", "ii"]
            )
        ):

            if phoneme == "an":
                final_phonemes.append("a")
            elif phoneme == "un":
                final_phonemes.append("u")
            elif phoneme == "in":
                final_phonemes.append("i")

            final_phonemes.append("mm")

        # apply ikfaa rules for noon and tanween .
        elif phoneme in ["n", "an", "un", "in"] and (
            (
                i + 3 < len(phonemes)
                and phonemes[i + 1] == SPACE_TOKEN
                and phonemes[i + 2] in IKFAA_LETTERS
                and phonemes[i + 3] in ["a", "u", "i", "aa", "uu", "ii"]
            )
            or (
                i + 2 < len(phonemes)
                and phonemes[i + 1] in IKFAA_LETTERS
                and phonemes[i + 2] in ["a", "u", "i", "aa", "uu", "ii"]
            )
        ):

            if phonemes[i + 1] == SPACE_TOKEN:
                final_phonemes.append(phoneme + phonemes[i + 2])
            else:
                final_phonemes.append(phoneme + phonemes[i + 1])

        else:
            final_phonemes.append(phoneme)

    return final_phonemes
```

```python
def apply_sukoon_meem_rules(phonemes: list) -> list:

    final_phonemes = []

    for i, phoneme in enumerate(phonemes):

        # apply ikfaa .
        if (
            phoneme == "m"
            and i + 2 < len(phonemes)
            and phonemes[i + 1] == SPACE_TOKEN
            and phonemes[i + 2] in IKFAA_SHAFAWI_LETTERS
        ):
            final_phonemes.append("mm")

        # apply idgham .
        elif (
            phoneme == "m"
            and i + 2 < len(phonemes)
            and phonemes[i + 1] == SPACE_TOKEN
            and phonemes[i + 2] in IDGHAM_SHAFAWI_LETTERS
        ):
            if phonemes[i + 2] != "mm":
                final_phonemes.append("mm")

        else:
            final_phonemes.append(phoneme)

    return final_phonemes
```

```python
def apply_idghamat_rules(phonemes: list) -> list:
    final_phonemes = []

    for i, phoneme in enumerate(phonemes):

        # apply idgham mutamathilayn rules.
        if (
            i + 3 < len(phonemes)
            and phoneme[0] == phonemes[i + 1][0]
            and (phonemes[i + 2][0] == phonemes[i + 1][0])
            and phonemes[i + 3] in ["a", "u", "i", "aa", "uu", "ii"]
            or (
                i + 4 < len(phonemes)
                and phonemes[i + 1] == SPACE_TOKEN
                and phoneme[0] == phonemes[i + 2][0]
                and phonemes[i + 2][0] == phonemes[i + 3][0]
                and phonemes[i + 4] in ["a", "u", "i", "aa", "uu", "ii"]
            )
        ):
            pass

        # apply idgham mutajanisen
        elif (
            i + 2 < len(phonemes)
            and (
                IDGHAM_MUTAJANISEN_LETTERS.get(phoneme) == phonemes[i + 1]
                or IDGHAM_MUTAJANISEN_LETTERS_2.get(phoneme) == phonemes[i + 1]
            )
            or (
                i + 3 < len(phonemes)
                and phonemes[i + 1] == SPACE_TOKEN
                and (
                    IDGHAM_MUTAJANISEN_LETTERS.get(phoneme) == phonemes[i + 2]
                    or IDGHAM_MUTAJANISEN_LETTERS_2.get(phoneme) == phonemes[i + 2]
                )
            )
        ):
            pass

        # apply idgham mutkarbeen
        elif (
            i + 2 < len(phonemes)
            and IDGHAM_MUTAKARBEEN_LETTERS.get(phoneme) == phonemes[i + 1]
            or (
                i + 3 < len(phonemes)
                and phonemes[i + 1] == SPACE_TOKEN
                and IDGHAM_MUTAKARBEEN_LETTERS.get(phoneme) == phonemes[i + 2]
            )
        ):
            pass

        else:
            final_phonemes.append(phoneme)

    return final_phonemes
```

```python
def apply_raa_rules(phonemes: list) -> list:

    final_phonemes = []

    for i, phoneme in enumerate(phonemes):

        if (
            phoneme == "r"
            and (
                i + 1 < len(phonemes)
                and phonemes[i + 1] in ["u", "a", "an", "un", "aa", "uu"]
            )
            or (
                phoneme == "r"
                # this mean raa has sukoon
                and (
                    (
                        i + 1 < len(phonemes)
                        and phonemes[i + 1]
                        not in ["a", "u", "i", "an", "un", "in", "aa", "uu", "ii"]
                        and phonemes[i + 1] != phoneme
                    )
                    or i == len(phonemes) - 1
                )
                and (
                    i - 1 >= 0
                    and phonemes[i - 1] != "y"
                    and (
                        phonemes[i - 1] in ["u", "a", "<wasl>"]
                        or (
                            phonemes[i - 1] in ["i"]
                            and i + 1 < len(phonemes)
                            and phonemes[i + 1] in ISTʕLAA_LETTERS
                            and (
                                i + 1 == len(phonemes) - 1
                                or phonemes[i + 2] not in ["i", "in", "ii"]
                            )
                        )
                    )
                    or (
                        i - 2 >= 0
                        and phonemes[i - 1] != "y"
                        and (
                            phonemes[i - 1] not in ["a", "u", "i"]
                            and phonemes[i - 2] in ["u", "a", "<wasl>"]
                            or (
                                phonemes[i - 2] in ["i"]
                                and i + 1 < len(phonemes)
                                and phonemes[i + 1] in ISTʕLAA_LETTERS
                                and (
                                    i + 1 == len(phonemes) - 1
                                    or phonemes[i + 2] not in ["i", "in", "ii"]
                                )
                            )
                        )
                    )
                )
            )
        ):
            final_phonemes.append("rM")
        else:
            final_phonemes.append(phoneme)

    return final_phonemes
```

```python
def apply_leen_madd(phonemes: list) -> list:
    final_phonemes = []
    for i, phoneme in enumerate(phonemes):
        if (
            phoneme == "y"
            and i + 1 < len(phonemes)
            and phonemes[i + 1] not in ["uu", "ii", "aa", "a", "u", "i", SILENT_TOKEN]
            and i - 1 >= 0
            and phonemes[i - 1] == "a"
            and (
                i + 1 == len(phonemes) - 1
                or (
                    i + 2 < len(phonemes)
                    and phonemes[i + 2]
                    not in ["a", "u", "i", "an", "un", "in", "aa", "uu", "ii"]
                )
            )
        ):
            final_phonemes.append(phoneme)
            final_phonemes.append("ii")
        elif (
            phoneme == "w"
            and i + 1 < len(phonemes)
            and phonemes[i + 1] not in ["uu", "ii", "aa", "a", "u", "i", SILENT_TOKEN]
            and i - 1 >= 0
            and phonemes[i - 1] == "a"
            and (
                i + 1 == len(phonemes) - 1
                or (
                    i + 2 < len(phonemes)
                    and phonemes[i + 2]
                    not in ["a", "u", "i", "an", "un", "in", "aa", "uu", "ii"]
                )
            )
        ):
            final_phonemes.append(phoneme)
            final_phonemes.append("uu")
        else:
            final_phonemes.append(phoneme)

    return final_phonemes
```

```python
def apply_qalqalaa_rules(phonemes: list) -> list:
    final_phonemes = []
    skip = False
    for i, phoneme in enumerate(phonemes):
        if skip:
            final_phonemes.append(phoneme + "K")
            skip = False
            continue
        if phoneme in QALQALAA_LETTERS and (
            (
                (
                    i + 1 < len(phonemes)
                    and phonemes[i + 1]
                    not in ["a", "i", "u", "ii", "uu", "aa", "an", "in", "un"]
                )
                or i == len(phonemes) - 1
            )
        ):
            if i + 1 < len(phonemes) and phonemes[i + 1] == phoneme:
                final_phonemes.append(phoneme)
                if (
                    i + 3 < len(phonemes)
                    and phonemes[i + 3] == SILENT_TOKEN
                    or i + 2 == len(phonemes) - 1
                ):
                    skip = True
            else:
                final_phonemes.append(phoneme + "K")
        else:
            final_phonemes.append(phoneme)
    return final_phonemes
```

```python
def apply_taa_marbuta_rules(phonemes: list) -> list:

    final_phonemes = []

    for i, phoneme in enumerate(phonemes):
        if phoneme == "T":
            if i == len(phonemes) - 1 or (
                i + 1 < len(phonemes) and phonemes[i + 1] == SILENT_TOKEN
            ):
                final_phonemes.append("h")
            else:
                final_phonemes.append("t")
        else:
            final_phonemes.append(phoneme)
    return final_phonemes
```

```python
def detect_edge_silence(wav_path, top_db=30, min_silence_ms=300):
    """
    Returns:
        has_start_silence (bool)
        has_end_silence (bool)
    """

    y, sr = librosa.load(wav_path, sr=16000, mono=True, res_type="soxr_hq")

    # non-silent intervals
    intervals = librosa.effects.split(y, top_db=top_db)

    if len(intervals) == 0:
        return True, True

    total_samples = len(y)

    first_start = intervals[0][0]
    last_end = intervals[-1][1]

    start_silence_ms = first_start / sr * 1000
    end_silence_ms = (total_samples - last_end) / sr * 1000

    has_start_silence = start_silence_ms >= min_silence_ms
    has_end_silence = end_silence_ms >= min_silence_ms

    return has_start_silence, has_end_silence
```

```python
def add_silent_char(phonemes: list, audio_path=None) -> list:

    if audio_path is not None:
        has_start_silence, has_end_silence = detect_edge_silence(audio_path)

        if has_start_silence:
            phonemes.insert(0, SILENT_TOKEN)

        if has_end_silence:
            phonemes.append(SILENT_TOKEN)

    return phonemes
```

```python
def apply_waqf_cases(phonemes: list, audio_path=None) -> list:
    final_phonemes = []
    skip = False

    phonemes = add_silent_char(phonemes, audio_path)

    for i, phoneme in enumerate(phonemes):

        if skip:
            skip = False
            continue

        if phoneme in ["a", "u", "i", "un", "in"] and (
            i == len(phonemes) - 1
            or (i + 1 < len(phonemes) and phonemes[i + 1] == SILENT_TOKEN)
        ):
            continue
        elif phoneme in ["an"] and (
            i == len(phonemes) - 1
            or (i + 1 < len(phonemes) and phonemes[i + 1] == SILENT_TOKEN)
        ):
            final_phonemes.append("aa")

        elif (
            phoneme in ["aa", "uu", "ii"]
            and i + 1 < len(phonemes)
            and phonemes[i + 1] == WASL
        ):
            final_phonemes.append(phoneme[0])

        elif phoneme == WASL:
            if (
                i == 0 or (i - 1 >= 0 and phonemes[i - 1] == SILENT_TOKEN)
            ) and i + 1 < len(phonemes):
                final_phonemes.append("ʔ")
            elif i + 1 < len(phonemes) and phonemes[i + 1] in ["a", "u", "i"]:
                skip = True
        else:
            final_phonemes.append(phoneme)

    return final_phonemes
```

```python
def remove_spaces(phonemes: list) -> list:
    return [p for p in phonemes if p != SPACE_TOKEN]
```

```python
def process_word(word, next_word):

    parsed = parse_arabic(word)

    next_word_parsed = parse_arabic(next_word)

    stem_text = parsed_to_string(parsed)

    # print(f"Processing word: {stem_text}")

    # -----------------------------------
    # SPECIAL WORD
    # -----------------------------------

    special = apply_special_word_override(stem_text)

    if special is not None:

        return special

    # -----------------------------------
    # ADD HAMZAT WASL TO HIDDEN WASL WORDS
    # -----------------------------------

    if stem_text in HIDDEN_HAMZA_WASL_WORDS:
        parsed.insert(1, (HAMZAT_WASL, []))

    # -----------------------------------
    # NORMAL WORD
    # -----------------------------------

    parsed = apply_hamzaat_wasl(parsed)

    parsed = remove_quranic_marks(parsed)
    parsed = apply_madd_silah(parsed, next_word_parsed)
    parsed = normalize_dagger_alif(parsed)
    parsed = normalize_alif_maqsura(parsed)
    parsed = apply_long_vowels(parsed)
    parsed = apply_al_rule(parsed)

    return arabic_to_phonemes(parsed)
```

```python
def text_to_phonemes(text: str, with_tajweed=True, audio_path=None) -> list:

    all_phonemes = []

    words = text.split()

    for i, word in enumerate(words):

        next_word = ""
        if i + 1 < len(words):
            next_word = words[i + 1]

        phonemes = process_word(word, next_word)

        if len(all_phonemes) > 0:
            all_phonemes.append(SPACE_TOKEN)
        all_phonemes.extend(phonemes)

    if with_tajweed:
        all_phonemes = apply_sukoon_noon_and_tanween_rules(all_phonemes)
        all_phonemes = apply_sukoon_meem_rules(all_phonemes)
        all_phonemes = apply_idghamat_rules(all_phonemes)

        all_phonemes = remove_spaces(all_phonemes)

        all_phonemes = apply_qalqalaa_rules(all_phonemes)
        all_phonemes = apply_waqf_cases(all_phonemes, audio_path)
        all_phonemes = apply_raa_rules(all_phonemes)
        all_phonemes = apply_leen_madd(all_phonemes)
        all_phonemes = apply_taa_marbuta_rules(all_phonemes)

    # TODO: apply duplicated words

    return all_phonemes
```
