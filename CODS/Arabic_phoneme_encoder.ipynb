```python
def remove_quranic_marks(parsed_text):
    """
    Remove all non-phonetic Quranic symbols.
    """

    return [(l, d) for l, d in parsed_text if l not in QURANIC_MARKS]
```

```python
def parse_arabic(text: str) -> List[Tuple[str, List[str]]]:
    """
    Parse Arabic text into (letter, diacritics) pairs.

    Example:
        "مِنْ" → [("م", ["ِ"]), ("ن", ["ْ"])]

    Rules:
    - Each base letter is grouped with its following diacritics
    - Diacritics are stored as a list
    - Spaces are preserved as (" ", [])
    - Quranic marks are ignored
    """

    result: List[Tuple[str, List[str]]] = []

    current_letter = None
    current_diacritics: List[str] = []

    for i, char in enumerate(text):

        # -------------------------------
        # 1. Skip Quranic marks
        # -------------------------------
        if char in QURANIC_MARKS:
            continue

        # ا۟  skip this letter like كَانُوا۟

        if (
            i + 1 < len(text)
            and char == "ا"
            and (text[i + 1] == "۟" or text[i + 1] == "۠")
        ):
            continue

        # -------------------------------
        # 2. Handle space
        # -------------------------------
        if char.isspace():
            # flush previous letter
            if current_letter is not None:
                result.append((current_letter, current_diacritics))
                current_letter = None
                current_diacritics = []

            result.append((" ", []))
            continue

        # -------------------------------
        # 3. If diacritic → attach it
        # -------------------------------
        if char in DIACRITICS:
            if current_letter is None:
                # edge case: diacritic without base letter
                continue

            current_diacritics.append(char)
            continue

        # -------------------------------
        # 4. New letter → flush previous
        # -------------------------------
        if current_letter is not None:
            result.append((current_letter, current_diacritics))

        current_letter = char
        current_diacritics = []

    # -------------------------------
    # 5. Flush last letter
    # -------------------------------
    if current_letter is not None:
        result.append((current_letter, current_diacritics))

    return result
```

```python
def is_madd_letter(letter, diacritics):
    """
    Check whether a letter can function as a madd letter.

    Madd letters:
    - ا
    - وْ
    - يْ

    They must not carry real vowels.
    """

    # Alif normally has no diacritics
    if letter == LONG_ALIF:
        return len(diacritics) == 0

    # Waw and Ya may carry sukoon
    if letter in [LONG_WAW, LONG_YA]:

        # Allowed:
        # []
        # [ْ]
        return len(diacritics) == 0 or diacritics == [SUKOON]

    return False


def is_word_final(parsed_text, i):
    j = i + 1
    while j < len(parsed_text):
        if parsed_text[j][0] == SPACE_TOKEN:
            return True

        if parsed_text[j][0].isalpha():
            return False
        j += 1
    return True


def parsed_to_string(parsed):

    text = ""

    for letter, diacritics in parsed:
        text += letter
        text += "".join(diacritics)

    return text


def has_vowel(diacritics):
    """
    Check whether a letter carries a real vowel.
    """

    return any(
        d
        in [
            FATHA,
            DAMMA,
            KASRA,
            FATHATAN,
            DAMMATAN,
            KASRATAN,
        ]
        for d in diacritics
    )
```

```python
def normalize_dagger_alif(parsed_text):
    """
    Convert dagger alif into LONG_ALIF marker.

    Example:
        مَـٰ → م + LONG_ALIF
    """

    result = []

    for letter, diacritics in parsed_text:

        if DAGGER_ALIF in diacritics or TATWEEL in diacritics:

            # remove dagger alif
            diacritics = [d for d in diacritics if d != DAGGER_ALIF]

            # add normalized long vowel marker
            diacritics.append("LONG_ALIF")

        result.append((letter, diacritics))

    return result
```

```python
def apply_special_word_override(word: str):
    """
    If word exists in SPECIAL_WORDS, return phonemes directly.
    """

    return SPECIAL_WORDS.get(word, None)
```

```python
def normalize_alif_maqsura(parsed_text):
    """
    Resolve final alif maqsurah (ى).

    Rules:

    1. Previous fatha + no vowel on ى
       -> LONG_ALIF

    2. Previous kasra + no vowel on ى
       -> LONG_YA

    3. ىً
       -> tanween fatḥ ("an")

    4. Otherwise
       -> consonant yaa
    """

    result = []

    for i, (letter, diacritics) in enumerate(parsed_text):

        # -----------------------------------
        # Normal letters
        # -----------------------------------
        if letter != "ى":
            result.append((letter, diacritics))
            continue

        # -----------------------------------
        # Need previous letter
        # -----------------------------------
        if len(result) == 0:
            result.append(("ي", diacritics))
            continue

        prev_letter, prev_diacritics = result[-1]

        # -----------------------------------
        # Case 1:
        # ىً -> an
        # -----------------------------------
        if FATHATAN in diacritics:

            prev_diacritics.append(FATHATAN)
            result[-1] = (prev_letter, prev_diacritics)

            continue

        # -----------------------------------
        # Detect whether current ى
        # has real vowel
        # -----------------------------------
        has_real_vowel = any(
            d in diacritics
            for d in [
                FATHA,
                DAMMA,
                KASRA,
                DAMMATAN,
                KASRATAN,
            ]
        )

        # -----------------------------------
        # Case 2:
        # previous fatha -> aa
        # -----------------------------------
        if LONG_ALIF_MARK in diacritics or (
            FATHA in prev_diacritics
            and not has_real_vowel
            and is_word_final(parsed_text, i)
            and SUKOON not in diacritics
        ):

            prev_diacritics.append(LONG_ALIF_MARK)
            result[-1] = (prev_letter, prev_diacritics)

            continue

        # -----------------------------------
        # Case 3:
        # previous kasra -> ii
        # -----------------------------------
        if (
            KASRA in prev_diacritics
            and not has_real_vowel
            and is_word_final(parsed_text, i)
        ):

            prev_diacritics.append(LONG_YA_MARK)
            result[-1] = (prev_letter, prev_diacritics)

            continue

        # -----------------------------------
        # Otherwise:
        # treat as consonant yaa
        # -----------------------------------
        result.append(("ي", diacritics))

    return result
```

```python
def apply_long_vowels(parsed_text):
    """
    Modify parsed text to merge long vowel patterns.

    Input:
        [("ق", ["َ"]), ("ا", []), ("ل", ["َ"])]

    Output:
        [("ق", ["َ", "LONG_ALIF"]), ("ل", ["َ"])]

    Then Step 2 will convert it to:
        ["q", "aa", "l", "a"]
    """

    result = []
    i = 0

    while i < len(parsed_text):
        letter, diacritics = parsed_text[i]

        # ----------------------------------
        # Look ahead for long vowel patterns
        # ----------------------------------
        if i + 1 < len(parsed_text):
            next_letter, next_diacritics = parsed_text[i + 1]

            # ------------------------------
            # 1. Fatha + Alif → aa
            # ------------------------------
            if (
                FATHA in diacritics
                and next_letter == LONG_ALIF
                and is_madd_letter(next_letter, next_diacritics)
            ):
                result.append((letter, diacritics + ["LONG_ALIF"]))
                i += 2
                continue

            # ------------------------------
            # 2. Kasra + Ya → ii
            # ------------------------------
            if (
                KASRA in diacritics
                and next_letter == LONG_YA
                and is_madd_letter(next_letter, next_diacritics)
            ):
                result.append((letter, diacritics + ["LONG_YA"]))
                i += 2
                continue

            # ------------------------------
            # 3. Damma + Waw → uu
            # ------------------------------
            if (
                DAMMA in diacritics
                and next_letter == LONG_WAW
                and is_madd_letter(next_letter, next_diacritics)
            ):
                result.append((letter, diacritics + ["LONG_WAW"]))
                i += 2
                continue

        # ----------------------------------
        # No long vowel → keep as is
        # ----------------------------------
        result.append((letter, diacritics))
        i += 1

    return result
```

```python
def apply_al_rule(parsed_text):
    """
    Apply sun/moon lam rules.

    Sun letters:
        lam disappears
        next letter gets shadda

    Moon letters:
        no change
    """

    result = []
    i = 0

    while i < len(parsed_text):

        # Need at least 3 chars:
        # ٱ + ل + next letter
        if i + 2 < len(parsed_text):

            first_letter, first_diacritics = parsed_text[i]
            second_letter, second_diacritics = parsed_text[i + 1]
            third_letter, third_diacritics = parsed_text[i + 2]

            # -----------------------------------
            # Detect الـ
            # -----------------------------------
            if (
                first_letter == HAMZAT_WASL
                and second_letter == "ل"
                and SUKOON not in second_diacritics
            ):

                # -----------------------------------
                # Sun letter case
                # -----------------------------------
                if third_letter in SUN_LETTERS:

                    # add shadda if missing
                    if SHADDA not in third_diacritics:
                        third_diacritics = [SHADDA] + third_diacritics

                    result.append((first_letter, first_diacritics))
                    result.append((third_letter, third_diacritics))

                    i += 3
                    continue

        # -----------------------------------
        # default
        # -----------------------------------
        result.append(parsed_text[i])
        i += 1

    return result
```

```python
def apply_madd_silah(parsed_text, next_word):
    """
    Apply Madd Silah Sughra.

    هُ → huu
    هِ → hii

    when followed by a vowelized letter.
    """

    result = []
    prev_diac = None

    str_text = parsed_to_string(parsed_text)

    #     Processing word: فِيهِ
    #     Processing word: مُهَانًا

    for i, (letter, diacritics) in enumerate(parsed_text):

        # -----------------------------------
        # Must have next letter
        # -----------------------------------

        if len(next_word) > 0:

            _, next_diacritics = next_word[0]

            next_text = parsed_to_string(next_word)

            if (
                has_vowel(next_diacritics)
                and is_word_final(parsed_text, i)
                and prev_diac is not None
                and len(prev_diac) > 0
                and SUKOON not in prev_diac
            ):

                # -----------------------------------
                # هُ → huu
                # -----------------------------------
                if (
                    letter == "ه"
                    and DAMMA in diacritics
                    and (str_text != "يَرْضَهُ" and next_text != "لَكُمْ")
                ):

                    diacritics = diacritics + ["LONG_WAW"]

                # -----------------------------------
                # هِ → hii
                # -----------------------------------
                elif letter == "ه" and KASRA in diacritics:

                    diacritics = diacritics + ["LONG_YA"]

            elif letter == "ه" and str_text == "فِيهِ" and next_text == "مُهَانًا":
                diacritics = diacritics + ["LONG_YA"]

        result.append((letter, diacritics))

        prev_diac = diacritics

    return result
```

```python
def arabic_to_phonemes(parsed_text):
    """
    Convert parsed Arabic [(letter, diacritics)] → list of phonemes.

    Output:
        ["m", "i", "n", "b", "a", "ʕ", "d", "i"]
    """

    phonemes = []
    word_closed = False

    for letter, diacritics in parsed_text:

        if word_closed:
            continue

        # -------------------------------
        # 1. Handle space
        # -------------------------------
        if letter == " ":

            phonemes.append(SPACE_TOKEN)
            word_closed = False
            continue

        # -------------------------------
        # 2. Get base consonant
        # -------------------------------
        base = ARABIC_TO_PHONEME.get(letter)

        if base is None:
            # unknown character → skip
            continue

        # -------------------------------
        # 3. Check shadda
        # -------------------------------
        has_shadda = SHADDA in diacritics

        if has_shadda:
            # double consonant
            phonemes.append(base)

        # always append base once
        phonemes.append(base)

        # -------------------------------
        # Detect long vowel first
        # -------------------------------
        has_long_alif = LONG_ALIF_MARK in diacritics
        has_long_ya = LONG_YA_MARK in diacritics
        has_long_waw = LONG_WAW_MARK in diacritics

        # -------------------------------
        # Emit vowel
        # -------------------------------
        for d in diacritics:

            # 🔴 Skip short vowel if long exists
            if has_long_alif and d == FATHA:
                continue

            if has_long_ya and d == KASRA:
                continue

            if has_long_waw and d == DAMMA:
                continue

            # -------------------------------
            # Normal handling
            # -------------------------------
            if d == FATHA:
                phonemes.append("a")

            elif d == DAMMA:
                phonemes.append("u")

            elif d == KASRA:
                phonemes.append("i")

            elif d == FATHATAN:
                phonemes.append("an")
                word_closed = True

            elif d == DAMMATAN:
                phonemes.append("un")
                word_closed = True

            elif d == KASRATAN:
                phonemes.append("in")
                word_closed = True

            elif d == LONG_ALIF_MARK:
                phonemes.append("aa")

            elif d == LONG_YA_MARK:
                phonemes.append("ii")

            elif d == LONG_WAW_MARK:
                phonemes.append("uu")

            elif d == SUKOON:
                pass

            elif d == SHADDA:
                pass

    return phonemes
```

```python
def process_word(word, next_word):

    parsed = parse_arabic(word)

    next_word_parsed = parse_arabic(next_word)

    stem_text = parsed_to_string(parsed)

    # print(f"Processing word: {stem_text}")

    # -----------------------------------
    # SPECIAL WORD
    # -----------------------------------

    special = apply_special_word_override(stem_text)

    if special is not None:

        return special
    # -----------------------------------
    # NORMAL WORD
    # -----------------------------------

    parsed = remove_quranic_marks(parsed)
    parsed = apply_madd_silah(parsed, next_word_parsed)
    parsed = normalize_dagger_alif(parsed)
    parsed = normalize_alif_maqsura(parsed)
    parsed = apply_long_vowels(parsed)
    parsed = apply_al_rule(parsed)

    return arabic_to_phonemes(parsed)
```

```python
def text_to_phonemes(text: str):

    all_phonemes = []

    words = text.split()

    for i, word in enumerate(words):

        next_word = ""
        if i + 1 < len(words):
            next_word = words[i + 1]

        phonemes = process_word(word, next_word)

        if len(all_phonemes) > 0:

            all_phonemes.append(SPACE_TOKEN)
        all_phonemes.extend(phonemes)

    return all_phonemes
```
